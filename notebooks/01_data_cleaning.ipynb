```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../data/processed/hotel_bookings.csv")
print("data loaded successfully")
```

```python
#Data Cleaning & Feature Engineering

# 1. Convert date
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])

# 2. Fill children
df['children'] = df['children'].fillna(0).astype(int)

# 3. Fill country
df['country'] = df['country'].fillna('Unknown')

# 4. Handle agent
df['agent'] = df['agent'].fillna(0)

# 5. Extra useful features
df['total_guests'] = (df['adults'] + df['children'] + df['babies'])

df['total_stay_nights'] = (df['stays_in_weekend_nights'] + df['stays_in_week_nights'])

df['total_previous_bookings'] = (df['previous_cancellations'] + df['previous_bookings_not_canceled'])
```

```python
df.info()
```

```python
print(df.columns.tolist())
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().mean() * 100).round(2)
})

missing_summary = (
    missing_summary[missing_summary["Missing Values"] > 0]
    .sort_values("Missing %", ascending=False)
)

missing_summary
```

```python
# The 'company' column contains 94.31% missing values.
# Since this project focuses on booking behavior, cancellations, pricing,
# and customer/channel analysis, the column provides limited analytical value.
# Therefore, it is removed to avoid introducing unreliable assumptions during analysis.

df = df.drop(columns="company")
```

```python
print(f"Columns after removing company: {df.shape[1]}")
```

```python
#Number of Duplicate rows in the dataset

duplicate_count = df.duplicated().sum()
print(f"Duplicate rows: {duplicate_count:,}")
```

```python
# Remove rows that are exact duplicates of another record

df = df.drop_duplicates().copy()
print(f"Rows after removing duplicates: {len(df):,}")
```

```python
print(f"Remaining duplicate rows: {df.duplicated().sum():,}")
print("duplicates handled")
```

```python
# Check whether guest counts contain negative values
print("Adults below 0:", (df["adults"] < 0).sum())
print("Children below 0:", (df["children"] < 0).sum())
print("Babies below 0:", (df["babies"] < 0).sum())
```

```python
# Check for bookings where no guests are recorded
zero_guest_bookings = df[
    (df["adults"] == 0) &
    (df["children"] == 0) &
    (df["babies"] == 0)
]

print(f"Bookings with zero guests: {len(zero_guest_bookings):,}")
```

```python
df_clean = df[df["total_guests"] > 0].copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))
```

```python
# Verify that no zero-guest bookings remain
print("Zero-guest bookings remaining:", (df_clean["total_guests"] == 0).sum())
```

```python
print("total_guests <= 0:", (df_clean["total_guests"] <= 0).sum())
print("total_stay_nights < 0:", (df_clean["total_stay_nights"] < 0).sum())
print("total_previous_bookings < 0:", (df_clean["total_previous_bookings"] < 0).sum())
print("ADR < 0:", (df_clean["adr"] < 0).sum())
```
